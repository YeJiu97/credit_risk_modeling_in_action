## 1.1 变量的命名与赋值

在 Python 中，变量就像是一个贴了标签的盒子，我们可以把数据放进去，随时取用。给变量命名是编程的第一项规范，尤其在风控建模这种需要高度可维护性的工程中，清晰的命名至关重要。
Python 的变量命名必须遵循以下原则：只能包含字母、数字和下划线，且不能以数字开头；不能使用 Python 的关键字（如 if, for, class 等）；区分大小写。在风控场景中，我们强烈建议使用“见名知意”的命名方式，采用小写字母加下划线的风格（蛇形命名法）。例如，表示用户年龄的变量应命名为 user_age 而不是 a，表示逾期天数的变量应命名为 days_overdue 而不是 d。这样的命名习惯能让其他同事（或者三个月后的你自己）一眼就能看懂代码的业务含义。
赋值操作使用等号 = 完成，这与数学中的“等于”含义不同，它表示将右边的数据“放入”左边的变量盒子里。

赋值操作使用等号 = 完成，这与数学中的“等于”含义不同，它表示将右边的数据“放入”左边的变量盒子里。

````python
# 信贷风控场景下的变量命名与赋值示例

# 定义一个用户的唯一标识符
user_id = "U202603048899"

# 定义用户的年龄（整数）
user_age = 28

# 定义用户的信用评分（浮点数，通常范围300-900）
credit_score = 655.5

# 定义该用户当前是否处于黑名单中（布尔值）
is_in_blacklist = False

# 定义用户的职业类别（字符串）
occupation = "Software_Engineer"

# 打印变量内容，验证赋值是否成功
print(f"用户 {user_id} 的年龄为 {user_age}，信用分为 {credit_score}")
print(f"是否在黑名单：{is_in_blacklist}")
````

在上述代码中，我们创建了五个变量，分别存储了不同类型的信息。注意最后一行使用的 f"..." 语法，这是 Python 的 f-string 格式化字符串，允许我们直接在字符串中嵌入变量的值，这在生成风控日志或报告时非常常用。

## 1.2 数值类型：整数与浮点数

在风控数据中，数值是最常见的数据类型。Python 将数值主要分为两类：整数 (int) 和 浮点数 (float)。

整数 (int) 用于表示没有小数部分的数字。在风控场景中，用户的年龄、家庭人口数、过去一个月的通话次数、逾期期数等，通常都是整数。整数运算精确，不会产生精度误差。

浮点数 (float) 用于表示带有小数部分的数字。涉及金额的字段，如贷款本金、每月还款额、利息、以及模型输出的违约概率（如 0.035），通常都是浮点数。这里需要特别警惕的是计算机中浮点数的精度问题。由于计算机底层二进制存储的限制，浮点数运算有时会出现微小的误差（例如 0.1 + 0.2 可能不等于精确的 0.3）。在处理涉及核心资金结算的风控逻辑时，如果精度要求极高，通常会使用专门的 decimal 模块，但在一般的特征工程和模型训练阶段，标准的 float 类型已经足够。

````python
# 数值类型的实际操作

# 整数示例：计算用户的总负债期数
current_loan_terms = 3      # 当前在贷期数
historical_paid_terms = 12  # 历史已结清期数
total_terms = current_loan_terms + historical_paid_terms

print(f"用户总借贷期数：{total_terms}")  # 输出：15

# 浮点数示例：计算预估月供
loan_amount = 50000.0       # 贷款金额
annual_rate = 0.06          # 年利率 6%
monthly_rate = annual_rate / 12
months = 24                 # 分24期

# 简单的等额本息估算公式 (仅为演示，非精确金融公式)
estimated_monthly_payment = (loan_amount * monthly_rate) / (1 - (1 + monthly_rate) ** (-months))

print(f"预估月供金额：{estimated_monthly_payment:.2f} 元") 
# :.2f 表示保留两位小数，符合金额展示习惯

# 类型转换示例
# 有时候数据源传来的年龄可能是字符串 "28"，需要转为整数才能比较
age_str = "28"
age_int = int(age_str) 

if age_int < 18:
    print("拒绝准入：未成年")
else:
    print("年龄校验通过")
````

在上面的代码中，我们展示了整数的加减运算和浮点数的复杂运算。特别注意 :.2f 的用法，这是格式化输出浮点数的标准技巧，确保金额显示为两位小数，符合金融行业的规范。同时，我们也演示了 int() 函数，用于将字符串类型的数字转换为真正的整数，这是数据清洗中极其常见的一步。

## 1.3 字符串：处理文本信息

字符串 (str) 是由单引号 ' ' 或双引号 " " 包裹起来的文本序列。在风控领域，虽然核心模型多基于数值，但大量的辅助信息都是文本形式，例如用户的身份证号、手机号、居住地址、职业描述以及设备指纹等。

Python 提供了丰富的字符串操作方法，其中在风控中最常用的包括：切片（提取子串）、拼接（组合信息）和类型转换。
切片操作允许我们提取字符串的一部分。语法是 string[start:end]，其中 start 是起始索引（从0开始），end 是结束索引（不包含该位置）。例如，中国的身份证号第7到14位代表出生日期，前6位代表地区码，我们可以通过切片轻松提取这些信息作为特征。

````python
# 字符串在风控中的应用

# 模拟一个用户的身份证号
id_card = "11010519900304289X"

# 切片操作：提取地区码 (前6位)
region_code = id_card[0:6]
print(f"用户地区码：{region_code}")

# 切片操作：提取出生年份 (第7位到第10位，索引6到10)
birth_year = id_card[6:10]
print(f"用户出生年份：{birth_year}")

# 动态生成风险标签
risk_level = "High"
user_tag = f"RISK_{risk_level}_{region_code}"
print(f"生成的风险标签：{user_tag}")

# 字符串方法：检查是否包含敏感词
job_description = "Freelance gambler"
if "gambler" in job_description.lower(): # .lower() 将字符串转为小写，防止大小写干扰
    print("警告：职业描述包含敏感词")
else:
    print("职业描述正常")
````

## 1.4 布尔值：风控规则的逻辑核心

布尔类型 (bool) 只有两个值：True（真）和 False（假）。它是计算机逻辑判断的基础，也是风控规则引擎的核心。任何一个风控规则，最终都会归结为一个布尔值：通过（True）或拒绝（False）。

布尔值通常不直接手动输入，而是通过比较运算产生。常见的比较运算符包括：等于 ==、不等于 !=、大于 >、小于 <、大于等于 >=、小于等于 <=。

此外，我们需要使用逻辑运算符将多个布尔条件组合起来：
- and：与运算，只有当两边都为 True 时，结果才为 True。
- or：或运算，只要有一边为 True，结果就为 True。
- not：非运算，将 True 变为 False，反之亦然。

````python
# 布尔值与逻辑运算在风控规则中的应用

# 基础数据
user_age = 22
credit_score = 580
has_overdue = True  # 当前是否有逾期
loan_amount = 20000

# 规则1：准入规则 - 年龄必须大于等于18 且 信用分大于600
# 注意：这里使用了 and，两个条件必须同时满足
rule_access = (user_age >= 18) and (credit_score > 600)
print(f"是否满足准入规则：{rule_access}") 
# 结果为 False，因为信用分580不大于600

# 规则2：拦截规则 - 如果有当前逾期 或者 信用分低于550，则直接拒绝
# 注意：这里使用了 or，只要命中一个条件即为真
rule_reject = has_overdue or (credit_score < 550)
print(f"是否触发拦截规则：{rule_reject}")
# 结果为 True，因为有当前逾期

# 规则3：人工审核规则 - 不满足自动通过，也不满足直接拒绝的，转人工
# 使用 not 对结果取反
need_manual_review = not rule_access and not rule_reject
print(f"是否需要人工审核：{need_manual_review}")
# 结果为 False，因为已经触发了拦截规则

# 复杂逻辑示例：大额贷款特殊规则
# 如果贷款金额大于1万，要求信用分必须大于650；否则只需大于600
high_amount_requirement = loan_amount > 10000
score_check = credit_score > 650 if high_amount_requirement else credit_score > 600

print(f"大额贷款特殊校验结果：{score_check}")
````

这段代码完整演示了如何利用布尔值构建一个简单的风控决策流。变量 rule_access 和 rule_reject 存储了判断结果。特别注意三元表达式 A if condition else B 的用法，这是一种简洁的写法，意为“如果条件成立则取A，否则取B”，在设定动态阈值时非常有用。

## 1.5 空值 None：处理数据缺失

在真实的风控数据中，缺失值是常态。用户可能没有填写职业，或者征信报告查无此人。Python 使用一个特殊的对象 None 来表示“无”或“空”。
需要严格区分的是：None 不等于 0，不等于空字符串 ""，也不等于 False。

- 0 是一个具体的数值，代表数量为零。
- "" 是一个长度为0的字符串，代表内容为空。
- False 代表逻辑上的假。
- None 代表“什么都没有”，“未知”。

在处理风控数据时，如果不正确处理 None，程序往往会报错（例如尝试对 None 进行数学运算）。因此，在进行任何计算或判断前，检查变量是否为 None 是一个良好的编程习惯。

````python
# 处理缺失值 None

# 假设某用户的年收入数据缺失
annual_income = None
debt_ratio_threshold = 0.5

# 错误的做法：直接参与运算，这会抛出 TypeError
# calculated_debt = loan_amount / annual_income  # 程序会在这里崩溃

# 正确的做法：先判断是否为 None
if annual_income is None:
    print("警告：收入数据缺失，无法计算负债收入比，转入人工审核流程。")
    decision = "Manual_Review"
else:
    # 只有当数据存在时才计算
    debt_ratio = loan_amount / annual_income
    if debt_ratio > debt_ratio_threshold:
        decision = "Reject"
    else:
        decision = "Pass"
    print(f"负债收入比：{debt_ratio}, 决策：{decision}")

# 使用 is None 进行比较是 Python 的标准规范
# 不要使用 == None，虽然在某些情况下有效，但 is None 更准确且高效
if annual_income == None: 
    pass # 不推荐写法

if annual_income is None:
    pass # 推荐写法
````

通过上述代码，我们看到了 None 在实际业务中的保护作用。is None 是判断变量是否为空的标准语法。在构建特征工程管道时，我们通常会先将 None 填充为默认值（如 0 或平均值），或者直接标记为“缺失”这一类特殊的类别，而不是让程序崩溃。
